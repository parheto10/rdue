```python
import sqlite3
```

```python
# Connexion à une base (crée le fichier s'il n'existe pas)
conn_old = sqlite3.connect("db_mondelez_last_old.sqlite3")
conn_last = sqlite3.connect("db_mondelez_last.sqlite3")

# Crée un curseur pour exécuter des requêtes SQL
cursor_old = conn_old.cursor()
cursor_last = conn_last.cursor()
```

```python
cursor_old.execute("SELECT * FROM myapi_risquerdue")
colonnes = [description[0] for description in cursor_old.description]
rows = cursor_old.fetchall()
print(len(rows))
print(colonnes)
```

```python
cursor_last.executemany(f"INSERT INTO myapi_risquerdue ({', '.join(colonnes)}) VALUES ({', '.join(['?' for _ in colonnes])})", rows)
conn_last.commit()
```
